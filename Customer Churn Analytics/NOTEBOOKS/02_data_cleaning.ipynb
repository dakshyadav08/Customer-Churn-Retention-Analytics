## Notebook 02: Data Cleaning

### Objective

The purpose of this notebook is to clean the raw telecom dataset before loading it into PostgreSQL and performing business analysis.

Tasks:
- Load raw data
- Inspect missing values
- Clean missing values
- Validate data
- Save cleaned dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\MSII\OneDrive\Desktop\Customer Churn Analytics\telco.csv")
```

```python
clean_df = df.copy()
```

```python
clean_df.shape
```

```python
missing = clean_df.isnull().sum()

missing[missing > 0]
```

```python
clean_df["Offer"] = clean_df["Offer"].fillna("No Offer")
```

```python
clean_df["Offer"].isnull().sum()
```

```python
clean_df["Internet Type"] = clean_df["Internet Type"].fillna("No Internet")
```

```python
clean_df["Internet Type"].isnull().sum()
```

```python
clean_df["Churn Category"] = clean_df["Churn Category"].fillna("Not Churned")
```

```python
clean_df["Churn Category"].isnull().sum()
```

```python
clean_df["Churn Reason"] = clean_df["Churn Reason"].fillna("Not Churned")
```

```python
clean_df["Churn Reason"].isnull().sum()
```

```python
clean_df[["Offer", "Internet Type", "Churn Category", "Churn Reason"]].isnull().sum()
```

```python
clean_df.to_csv(
    r"C:\Users\MSII\OneDrive\Desktop\Customer Churn Analytics\telco_cleaned.csv",
    index=False
)
```
